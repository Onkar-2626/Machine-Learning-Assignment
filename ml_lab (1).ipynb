```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
print('Libraries imported.')
```

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/glass/glass.data"
columns = ['id','RI','Na','Mg','Al','Si','K','Ca','Ba','Fe','class']
df = pd.read_csv(url, names=columns)
df.head()
```

```python

X = df.iloc[:, 0:4]
y = df['id']
print(X.shape, y.shape)
```

```python

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42)

print('Train size:', X_train.shape)
print('Test size:', X_test.shape)
```

```python

knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)
print('Model trained.')
```

```python


y_pred = knn.predict(X_test)
list(y_pred)[:10]
```

```python
results = pd.DataFrame({
    'Actual': y_test.values,
    'Predicted': y_pred
})

results['Status'] = results.apply(lambda r: 'Correct' if r.Actual == r.Predicted else 'Wrong', axis=1)
results.head(20)
```

```python

print('Accuracy:', accuracy_score(y_test, y_pred))
print('\nClassification Report:')
print(classification_report(y_test, y_pred))
print('\nConfusion Matrix:')
print(confusion_matrix(y_test, y_pred))
```

In this assignment, the k-Nearest Neighbour (kNN) algorithm was successfully implemented to perform classification on a dataset similar to the Iris dataset. The dataset was first explored and divided into training and testing subsets to evaluate the model’s performance on unseen data.

Using k = 3, the kNN classifier was able to correctly classify most of the test samples, achieving a high accuracy. The confusion matrix and classification report further confirmed that the model performed well across all classes, with only a small number of misclassifications. This indicates that kNN is effective for datasets with well-separated classes and numerical features.
